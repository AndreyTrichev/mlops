```python
conn = psycopg2.connect(
    host="db",
    database="trichev03",
    user="andrey",
    password="trichev"
)
cur = conn.cursor()
```

```python
cur.execute("DROP TABLE IF EXISTS customers")

create_table_query = """
CREATE TABLE customers (
    id SERIAL PRIMARY KEY,
    customer_name TEXT,
    age INT,
    city TEXT,
    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);
"""

cur.execute(create_table_query)
conn.commit()
```

```python

insert_query = """
INSERT INTO customers (customer_name, age, city)
VALUES (%s, %s, %s)
"""

customers_data = [
    ('Иван Петров', 28, 'Москва'),
    ('Мария Сидорова', 34, 'Санкт-Петербург'),
    ('Алексей Иванов', 42, 'Казань')
]

for customer in customers_data:
    cur.execute(insert_query, customer)

conn.commit()
```
